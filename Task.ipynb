## OUTLIER DETECTION

```python
num_cols = df.select_dtypes(include= "number").columns
print(num_cols)
```

```python
num_cols = df.select_dtypes(include="number").columns

df[num_cols].skew().sort_values(ascending=False)
```

All the cols are approximately symmetrical as they have |skewness| < 0.5<br>


### OUTLIER DETECTION Using IQR Method

```python
Q1 = df[num_cols].quantile(0.25)
Q3 = df[num_cols].quantile(0.75)
IQR = Q3 - Q1
lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR
outlier_count = ((df[num_cols] < lower) | (df[num_cols] > upper)).sum()
print(outlier_count)
```

By using the IQR method for outlier detection we have found out there are no outliers<br>
We can also confirm it by making boxplots also

```python
df[['Current_Stock', 'Profit_USD']].boxplot(figsize=(10, 6))
plt.title('Boxplot of Current Stock and Profit')
plt.ylabel('Values')
plt.show()
```

From the graph we can clearly see that there are no outliers here in these cols<br>
Similarly for other num cols also

## FEATURE ENGINEERING

```python
# Stock_Cover_Ratio

df['Stock_Cover_Ratio'] = df['Current_Stock'] / df['Monthly_Demand'].replace(0,np.nan)
```

It tell usabout how much current stock is available in comparison with monthly demand<br>
High ratio- stock is relatively high<br>
Low ratio- stock is relatively low

```python
# Storage_Utilisation

df['Storage_Utilization'] = df['Current_Stock'] / df['Storage_Capacity']
```

It tells us about how much storage capacity of the warehouse is utilised by the current stock

```python
df[['Stock_Cover_Ratio', 'Storage_Utilization']].describe()
```

```python
df[['Stock_Cover_Ratio', 'Storage_Utilization']].info()
```

```python
df['Stock_Cover_Ratio'] = df['Stock_Cover_Ratio'].fillna(0)
```
